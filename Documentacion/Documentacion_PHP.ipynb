##  1. Configuración y Archivos

### 1.1 Pre-requisitos
Para ejecutar PHP, necesitas un **servidor web** local. Las opciones más comunes son:
* **XAMPP** (Windows, macOS, Linux)
* **WAMP** (Windows)
* **MAMP** (macOS)

**En VS Code:** Asegúrate de tener las extensiones **Jupyter**, **PHP Intelephense** y, si deseas probar los archivos rápidamente, **PHP Server** instaladas.

### 1.2 Creación del Archivo
El código PHP se guarda en archivos con la extensión **`.php`** (ej: `inicio.php`). Para que el servidor lo ejecute, el archivo debe estar en el directorio raíz del servidor (ej: la carpeta `htdocs` en XAMPP).

##  2. Sintaxis Básica

El código PHP siempre debe estar encerrado dentro de las etiquetas de apertura y cierre. El código fuera de estas etiquetas se trata como **HTML** simple.

* **Apertura:** `<?php`
* **Cierre:** `?>`

### 2.1 "Hola Mundo"
La instrucción más común para mostrar contenido en PHP es **`echo`** (o `print`). Cada instrucción debe terminar con un **punto y coma** (`;`).

```python
<?php

// Esto es un comentario de una sola línea
/*
    Esto es un bloque
    de comentario
*/

echo "¡Hola, mundo desde PHP!";

// Otra forma de usar echo es sin paréntesis, ya que es una construcción del lenguaje.
echo "<br>Esta línea se mostrará en una nueva línea HTML.";

?>
```

##  3. Variables y Tipos de Datos

### 3.1 Declaración de Variables
En PHP, todas las variables comienzan con el símbolo del **dólar (`$`)**. No es necesario declarar el tipo de dato, ya que PHP lo infiere (tipado dinámico).

| Tipo de Dato | Ejemplo | Descripción |
| :---: | :---: | :--- |
| **String** | `$nombre = "Juan";` | Cadenas de texto. |
| **Integer** | `$edad = 30;` | Números enteros. |
| **Float** | `$precio = 19.99;` | Números con decimales. |
| **Boolean** | `$es_valido = true;` | Valores lógicos (`true` o `false`). |

```python
<?php

$nombre = "Laura";
$edad = 28;
$saludo = "Hola, ";

// Concatenación: Se usa el punto (.) para unir strings.
echo $saludo . $nombre . ". Tienes " . $edad . " años.<br>";

// Interpolación: Las variables se interpretan dentro de comillas dobles ("")
echo "Tu nombre es $nombre y tu edad es $edad.";

?>
```

##  4. Arrays (Arreglos)

Los Arrays son colecciones de valores. PHP soporta **Arrays indexados** (con índices numéricos) y **Arrays asociativos** (con claves de texto).

### 4.1 Array Indexado

```python
<?php

$frutas = array("Manzana", "Banana", "Cereza");

// Acceso por índice (empieza en 0)
echo "Mi fruta favorita es la: " . $frutas[1] . "<br>"; // Muestra "Banana"

// Contar elementos
echo "Hay " . count($frutas) . " frutas en la lista.";

?>
```

### 4.2 Array Asociativo
Usan pares **clave/valor** para acceder a los datos.

```python
<?php

$persona = [
    'nombre' => 'María',
    'ciudad' => 'Madrid',
    'profesion' => 'Ingeniera'
];

// Acceso por clave (como un diccionario)
echo "La persona es " . $persona['nombre'] . " y trabaja como " . $persona['profesion'] . ".";

?>
```

##  5. Estructuras de Control

### 5.1 Condicionales (`if`, `elseif`, `else`)
Evalúan expresiones lógicas para decidir qué bloque de código ejecutar.

```python
<?php

$calificacion = 78;

if ($calificacion >= 90) {
    echo "¡Calificación A: Sobresaliente!";
} elseif ($calificacion >= 70) {
    echo "Calificación B: Notable.";
} else {
    echo "Calificación C: Necesitas repasar.";
}

?>
```

### 5.2 Bucles (`for` y `foreach`)

* **`for`**: Se usa cuando sabes cuántas veces necesitas iterar.
* **`foreach`**: La forma más común de iterar sobre un Array.

```python
<?php

// Bucle 'for'
echo "<h4>Números del 1 al 3:</h4>";
for ($i = 1; $i <= 3; $i++) {
    echo "Número: $i <br>";
}


// Bucle 'foreach' con Array Asociativo
$colores = ['primary' => 'Azul', 'secondary' => 'Gris'];
echo "<h4>Colores:</h4>";
foreach ($colores as $clave => $valor) {
    echo "Clave: $clave, Valor: $valor <br>";
}

?>
```

##  6. Funciones

Las funciones permiten encapsular y reutilizar bloques de código.

### 6.1 Definición y Llamada
Se utiliza la palabra clave **`function`** seguida del nombre y sus parámetros. Se usa `return` para devolver un valor.

```python
<?php

// Definición de la función con parámetros
function saludoFormal($nombre, $apellido)
{
    // Concatena y devuelve la cadena
    return "Estimado/a: " . $nombre . " " . $apellido;
}

// Llamada a la función
$mensaje = saludoFormal("Roberto", "Gómez");

echo $mensaje;

?>
```

##  7. Variables Superglobales

PHP tiene un conjunto de *Arrays* predefinidos llamados **Superglobales** que contienen información del entorno, del usuario, y de la solicitud.

La Superglobal **`$_SERVER`** es una de las más útiles, ya que contiene información del servidor web.

```python
<?php

// Muestra la dirección IP del usuario
echo "Tu dirección IP es: " . $_SERVER['REMOTE_ADDR'] . "<br>";

// Muestra el nombre del script actual
echo "El script que se está ejecutando es: " . $_SERVER['PHP_SELF'] . "<br>";

// Muestra el tipo de solicitud HTTP (GET, POST, etc.)
echo "Método de la Solicitud: " . $_SERVER['REQUEST_METHOD'];

?>
```
