##  2. Sintaxis Básica y Salida

### 2.1 Salida Estándar
En Ruby, la función principal para mostrar contenido es **`puts`** (que añade un salto de línea) o **`print`** (que no lo hace). Los comentarios se inician con el símbolo de la almohadilla (`#`).

```python
# Esto es un comentario de una sola línea

puts "¡Hola, mundo desde Ruby!"

# print no añade salto de línea automáticamente
print "El resultado es: "
puts 5 + 3

# Los bloques de código no usan punto y coma al final de las instrucciones.
```

##  3. Variables y Tipos de Datos

### 3.1 Declaración y Tipos
Las variables en Ruby no requieren un símbolo especial de prefijo (como el `$`). El lenguaje es de tipado dinámico.

* **Variables Locales:** Comienzan con una letra minúscula o guion bajo.
* **Constantes:** Comienzan con una letra mayúscula.

```python
nombre = "Elena"
edad = 35
pi_value = 3.14159
ESTADO = "ACTIVO"

# Concatenación simple (usando +)
puts "Nombre: " + nombre + ", Edad: " + edad.to_s

# Interpolación de cadenas (dentro de comillas dobles "") usando #{}
puts "Pi es aproximadamente #{pi_value}"

# Mostrar una constante
puts "El estado es: #{ESTADO}"
```

##  4. Estructuras de Datos (Arrays y Hashes)

### 4.1 Array (Arreglos)
Colecciones ordenadas e indexadas.

```python
frutas = ["Manzana", "Naranja", "Pera"]

# Acceso por índice (empieza en 0)
puts "La segunda fruta es: #{frutas[1]}"

# Añadir un elemento al final
frutas.push("Uva")
puts "Lista completa: #{frutas}"
```

### 4.2 Hash (Diccionario Asociativo)
Colecciones de pares clave/valor, similares a los diccionarios o mapas.

```python
persona = {
    "nombre" => "Carlos",
    "profesion" => "Arquitecto",
    :edad => 40 # El uso de símbolos (:) es la práctica moderna
}

# Acceso por clave
puts "Profesión de Carlos: #{persona['profesion']}"
puts "Edad: #{persona[:edad]}"
```

##  5. Estructuras de Control y Métodos

### 5.1 Condicionales (`if`, `elsif`, `else`)

```python
puntuacion = 85

if puntuacion >= 90
    puts "Excelente"
elsif puntuacion >= 70
    puts "Aceptable"
else
    puts "Necesitas esforzarte"
end
```

### 5.2 Definición de Métodos (Funciones)
En Ruby, las funciones se llaman **métodos** y se definen con la palabra clave **`def`** y se cierran con **`end`**. El valor devuelto es la última expresión evaluada, `return` es opcional.

```python
def calcular_area(ancho, alto)
    # No es necesario usar 'return' explícitamente, la última línea es el retorno
    ancho * alto
end

# Llamada al método
area_total = calcular_area(10, 5)

puts "El área calculada es: #{area_total}"
```
