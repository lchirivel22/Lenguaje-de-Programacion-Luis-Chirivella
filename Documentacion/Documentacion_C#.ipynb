# Introducción a C# (C Sharp)

Este Notebook documenta los fundamentos del lenguaje **C#**, un lenguaje de tipado fuerte y orientado a objetos, esencial para el desarrollo de aplicaciones con la plataforma .NET.

---

##  1. Configuración y Ejecución

### 1.1 Pre-requisitos
Para ejecutar C#:
* **SDK de .NET:** Debes tener el SDK de .NET instalado.
* **VS Code:** La extensión de Jupyter y el **Polyglot Notebooks** (anteriormente .NET Interactive) para poder ejecutar código C# directamente en las celdas del Notebook.

### 1.2 Estructura
C# es un lenguaje **tipado fuertemente**, lo que significa que las variables deben tener un tipo definido. El código se ejecuta tradicionalmente dentro de una función `Main` o, más comúnmente en proyectos modernos, usando **Top-level Statements**.

## 2. Sintaxis Básica y Salida

### 2.1 El método `Console.WriteLine`
La forma estándar de mostrar contenido en C# es usando el método **`Console.WriteLine()`** (para añadir un salto de línea) o `Console.Write()`. Todas las instrucciones deben terminar con un **punto y coma** (`;`).

```
// Esto es un comentario de una sola línea

Console.WriteLine("¡Hola, mundo desde C#!");

/*
    Este es un bloque
    de comentario
*/

Console.Write("La suma es: ");
Console.WriteLine(10 + 20);
```

##  3. Variables y Tipos de Datos

### 3.1 Tipos de Datos Primitivos
C# requiere que declares el tipo de dato de una variable antes de usarla (tipado estático). También puedes usar la palabra clave **`var`** para que el compilador infiera el tipo.

| Tipo de Dato | Ejemplo | Descripción |
| :---: | :---: | :--- |
| **string** | `string nombre = "Ana";` | Cadenas de texto. |
| **int** | `int edad = 24;` | Números enteros. |
| **double** | `double peso = 65.5;` | Números decimales de doble precisión. |
| **bool** | `bool activo = true;` | Valores lógicos (`true` o `false`). |

```
string nombre = "Miguel";
int unidades = 50;
double iva = 0.16;

// Concatenación usando el operador (+)
Console.WriteLine("Nombre: " + nombre + ", Unidades: " + unidades);

// Interpolación de cadenas (usando el símbolo $ y llaves {})
Console.WriteLine($"El IVA es del: {iva * 100}%");

// Uso de 'var' para inferir el tipo
var fecha = DateTime.Now;
Console.WriteLine($"Fecha actual: {fecha}");
```

##  4. Estructuras de Datos

### 4.1 Arrays
Colecciones de elementos del **mismo tipo** con un tamaño fijo.

```
// Declaración e inicialización de un array de strings de tamaño 3
string[] carros = new string[3];
carros[0] = "Toyota";
carros[1] = "Ford";
carros[2] = "BMW";

// Acceso por índice
Console.WriteLine($"El primer carro es: {carros[0]}");

// Obtener la longitud
Console.WriteLine($"El array tiene {carros.Length} elementos.");
```

### 4.2 Listas Genéricas (`List<T>`)
La forma preferida de manejar colecciones de tamaño dinámico.

```
// Es necesario usar 'using System.Collections.Generic;' para usar List
using System.Collections.Generic;

List<int> numeros = new List<int> { 10, 20, 30 };

// Añadir un elemento
numeros.Add(40);

Console.WriteLine($"El último número añadido es: {numeros[3]}");
```

##  5. Estructuras de Control

### 5.1 Condicionales (`if`, `else if`, `else`)

```
int temperatura = 15;

if (temperatura > 25)
{
    Console.WriteLine("Hace calor.");
}
else if (temperatura > 18)
{
    Console.WriteLine("Clima agradable.");
}
else
{
    Console.WriteLine("Hace frío.");
}
```

### 5.2 Bucles (`for` y `foreach`)

* **`for`**: Se usa para iterar un número conocido de veces.
* **`foreach`**: Se usa para iterar sobre los elementos de una colección.

```
// Bucle 'for'
Console.WriteLine("Cuenta progresiva:");
for (int i = 0; i < 3; i++)
{
    Console.WriteLine(i);
}

// Bucle 'foreach' con un Array
string[] dias = { "Lun", "Mar", "Mié" };
Console.WriteLine("\nDías de la semana:");
foreach (string dia in dias)
{
    Console.WriteLine(dia);
}
```

##  6. Métodos (Funciones)

En C#, las funciones se llaman **métodos**. Para que un método sea accesible desde la función principal del programa, a menudo se declara como **`static`**.

### 6.1 Definición y Llamada

```
// Definición de un método estático que devuelve un valor (int)
int Sumar(int a, int b)
{
    // C# es tipado fuerte, se debe especificar el tipo de retorno (int)
    return a + b;
}

// Llamada al método
int resultado = Sumar(15, 7);

Console.WriteLine($"La suma de los números es: {resultado}");
```
